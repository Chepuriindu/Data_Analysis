```python
import matplotlib.pyplot as plt
```

```python
df.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
```

```python
df.groupby("Category")["Profit"].sum().plot(kind="bar")
plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
df.groupby(df["Order_Date"].dt.month)["Sales"].sum().plot(kind="line", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

```python
df.groupby(df["Order_Date"].dt.month)["Sales"].sum().sort_values(ascending=False).head(3)
```

```python
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Units Sold:", df["Quantity"].sum())
print("Average Sales:", df["Sales"].mean())
print("Profit Margin (%):",
      (df["Profit"].sum() / df["Sales"].sum()) * 100)
```
