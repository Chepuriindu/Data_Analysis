```python
df_recent = df[df['year'] >= 2000].copy()
df_recent.head()
```

```python
df_recent.shape
```

```python
latest = df_recent[df_recent['year'] == 2024]
top10 = latest.nlargest(10, 'co2')[['country', 'co2']]
top10
```

```python
latest = latest[latest['iso_code'].notna()]
top10 = latest.nlargest(10, 'co2')[['country', 'co2']]
top10
```

```python
top10.plot(
    x='country',
    y='co2',
    kind='bar',
    figsize=(10,5),
    legend=False,
    title='Top 10 CO₂ Emitting Countries (2024)'
)
plt.ylabel('CO₂ emissions')
plt.xticks(rotation=45)
plt.show()
```

```python
world = df[df['country'] == 'World']

plt.figure(figsize=(12,5))
plt.plot(world['year'], world['co2'])
plt.title('Global CO₂ Emissions Over Time')
plt.xlabel('Year')
plt.ylabel('CO₂ emissions')
plt.show()
```

```python
latest[['cement_co2', 'coal_co2', 'oil_co2',
        'gas_co2', 'flaring_co2']].sum()
```

```python
latest[['cement_co2', 'coal_co2', 'oil_co2',
        'gas_co2', 'flaring_co2']].sum().plot(
    kind='bar', figsize=(8,5), title='CO₂ Emissions by Source (2024)'
)
plt.ylabel('CO₂ emissions')
plt.show()
```

```python
top_per_capita = latest.nlargest(10, 'co2_per_capita')[['country', 'co2_per_capita']]
top_per_capita
```

```python
countries = df_recent[df_recent['country'].isin(['India', 'China', 'United States'])]

for country in ['India', 'China', 'United States']:
    data = countries[countries['country'] == country]
    plt.plot(data['year'], data['co2'], label=country)

plt.title('CO₂ Emissions: India vs China vs USA')
plt.xlabel('Year')
plt.ylabel('CO₂ emissions')
plt.legend()
plt.show()
```

```python
countries.groupby('country')['co2'].mean().sort_values(ascending=False)
```

```python
countries.groupby(['country', 'year'])['co2'].sum().unstack().loc[:, [2000, 2024]]
```

```python
values = countries.groupby(['country', 'year'])['co2'].sum().unstack()

growth = ((values[2024] - values[2000]) / values[2000]) * 100
growth.sort_values(ascending=False)
```
